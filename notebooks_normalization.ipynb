```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# Create dataset
data = {
    'Name': ['Asha', 'Ravi', 'Priya', 'Kiran', 'Rahul'],
    'Age': [20, 25, 30, 35, 40],
    'Marks': [40, 50, 60, 70, 80],
    'Salary': [10000, 20000, 30000, 40000, 50000]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Create MinMaxScaler
scaler = MinMaxScaler()

# Normalize numerical columns
df[['Age', 'Marks', 'Salary']] = scaler.fit_transform(
    df[['Age', 'Marks', 'Salary']]
)

print("\nNormalized Dataset:")
print(df)
```
